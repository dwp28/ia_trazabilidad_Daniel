## Parte 3

```python
import hashlib, os

# Cambia el nombre del archivo por el de tu notebook real
nombre_archivo = "00_setup_WillsonDaniel.ipynb"  # <-- cámbialo

# Si aún no existe el archivo (primera ejecución), crea un texto provisional
if not os.path.exists(nombre_archivo):
    open(nombre_archivo, "w").write("Plantilla reproducible")

# Abre el archivo en modo binario y calcula el hash SHA-256
with open(nombre_archivo, "rb") as f:
    contenido = f.read()
    sha = hashlib.sha256(contenido).hexdigest()

print("SHA-256 generado:\n", sha)
```

## Parte 4

```python
import json, datetime

# Creamos el registro con los datos del hash y el nombre del archivo
registro = {
    "archivo": nombre_archivo,        # usa la variable del paso 3
    "sha256": sha,                    # hash calculado antes
    "fecha": datetime.datetime.now().isoformat()  # fecha actual ISO 8601
}

# Guardamos en un archivo JSON llamado "registro_hash.json"
with open("registro_hash.json", "w") as f:
    json.dump(registro, f, indent=2)

print("Registro creado: registro_hash.json ✅")
```
